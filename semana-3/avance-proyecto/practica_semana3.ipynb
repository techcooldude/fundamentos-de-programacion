## Ejercicio 1 — Costo de un paquete de sesiones (Temas 1, 2 y 3)
**Enunciado:** calcular el costo de *n* sesiones de fisioterapia a \$600 cada una.

**Modelo EPS**
- **Entrada:** `sesiones` (int), `precio` (float).
- **Proceso:** `total = precio * sesiones`.
- **Salida:** costo total con formato de moneda.

**Pseudocódigo (PSeInt)**
```
Algoritmo CostoPaquete
    Definir sesiones Como Entero
    Definir precio, total Como Real
    precio <- 600
    Leer sesiones
    total <- precio * sesiones
    Escribir "Total: $", total
FinAlgoritmo
```
**Procedimiento:** se declaran los tipos de dato, se aplica el operador aritmético `*` y se muestra el resultado con `print` usando un f-string con separador de miles.

````python
precio = 600.0          # float
sesiones = 4            # int  -> sesiones = int(input("Sesiones: "))
total = precio * sesiones
print(f"Costo de {sesiones} sesiones: ${total:,.2f}")
print(type(precio), type(sesiones), type(total))
````

## Ejercicio 2 — Casting y operadores (Temas 3 y 4)
**Enunciado:** leer la edad y el peso del paciente como texto, convertirlos y calcular el IMC.

**Modelo EPS**
- **Entrada:** `edad_txt`, `peso_txt`, `estatura_txt` (str).
- **Proceso:** casting con `int()` y `float()`; `imc = peso / estatura ** 2`; operadores `//` y `%` para años y meses.
- **Salida:** IMC redondeado y edad expresada en décadas y años.

**Procedimiento:** `input()` siempre devuelve `str`, por eso se convierte antes de operar. Se muestran operadores aritméticos (`/`, `**`, `//`, `%`).

````python
edad_txt = "47"        # input("Edad: ")
peso_txt = "72.5"      # input("Peso (kg): ")
estatura_txt = "1.68"  # input("Estatura (m): ")

edad = int(edad_txt)
peso = float(peso_txt)
estatura = float(estatura_txt)

imc = peso / estatura ** 2
print(f"IMC: {imc:.1f}")
print(f"{edad} años = {edad // 10} décadas y {edad % 10} años")
````

## Ejercicio 3 — Clasificación del dolor (Tema 5)
**Enunciado:** según el nivel de dolor (1-10), clasificar la atención.

**Modelo EPS**
- **Entrada:** `dolor` (int).
- **Proceso:** `if/elif/else` con operadores relacionales: 1-3 leve, 4-7 moderado, 8-10 prioritario.
- **Salida:** clasificación y booleano `prioritario`.

**Procedimiento:** las condiciones se evalúan de arriba hacia abajo; la primera verdadera determina la clasificación. Se prueba con varios valores usando una lista de casos.

````python
for dolor in (2, 5, 9, 11):
    if dolor < 1 or dolor > 10:
        clase = "Dato inválido"
    elif dolor <= 3:
        clase = "Leve"
    elif dolor <= 7:
        clase = "Moderado"
    else:
        clase = "Prioritario"
    prioritario = 8 <= dolor <= 10
    print(f"Dolor {dolor:2d}: {clase:<13} prioritario={prioritario}")
````

## Ejercicio 4 — Descuento no acumulable (Tema 5, operadores lógicos)
**Enunciado:** aplicar 15 % a adultos mayores (≥ 60) o 10 % a paquetes de 5+ sesiones; si cumple ambos se aplica solo el mayor.

**Modelo EPS**
- **Entrada:** `edad` (int), `sesiones` (int), `subtotal` (float).
- **Proceso:** booleanos `es_mayor`, `es_paquete`; decisión anidada.
- **Salida:** porcentaje, descuento y total.

**Procedimiento:** es la regla de negocio central del proyecto. Como 15 % > 10 %, se evalúa primero la condición de adulto mayor.

````python
casos = ((67, 6), (30, 6), (65, 2), (25, 1))
for edad, sesiones in casos:
    subtotal = 600.0 * sesiones
    es_mayor = edad >= 60
    es_paquete = sesiones >= 5
    if es_mayor:
        porcentaje = 0.15
    elif es_paquete:
        porcentaje = 0.10
    else:
        porcentaje = 0.0
    descuento = subtotal * porcentaje
    print(f"edad={edad}, sesiones={sesiones}, ambos={es_mayor and es_paquete}: "
          f"{porcentaje:.0%} -> total ${subtotal - descuento:,.2f}")
````

## Ejercicio 5 — Validación de entrada con `while` (Tema 6)
**Enunciado:** pedir el número de sesiones hasta que sea un entero entre 1 y 10.

**Modelo EPS**
- **Entrada:** respuestas sucesivas del usuario (str).
- **Proceso:** `while` que se repite mientras el dato **no** sea válido (`not isdigit()` o fuera de rango).
- **Salida:** valor válido y número de intentos.

**Procedimiento:** se simulan las respuestas con una secuencia; en el prototipo real cada vuelta llama a `input()`.

````python
respuestas = ["abc", "0", "15", "6"]   # simulación de input()
i = 0
dato = respuestas[i]
intentos = 1
while not dato.isdigit() or int(dato) < 1 or int(dato) > 10:
    print(f"'{dato}' no es válido, intenta de nuevo")
    i += 1
    dato = respuestas[i]
    intentos += 1
sesiones = int(dato)
print(f"Sesiones aceptadas: {sesiones} (en {intentos} intentos)")
````

## Ejercicio 6 — Menú con acumuladores (Tema 6)
**Enunciado:** simular la caja del día: registrar cobros hasta elegir "salir" y acumular totales por forma de pago.

**Modelo EPS**
- **Entrada:** opción de menú y monto/forma de pago.
- **Proceso:** `while opcion != "3"` con contadores y acumuladores (`+=`).
- **Salida:** corte de caja: total efectivo, tarjeta, número de cobros y promedio.

**Procedimiento:** es la estructura del menú principal del prototipo, reducida a lo esencial.

````python
operaciones = [("1", 600.0, "E"), ("1", 4248.75, "T"), ("2", 0, ""), ("1", 1000.0, "E"), ("3", 0, "")]
efectivo = 0.0
tarjeta = 0.0
cobros = 0
k = 0
opcion = ""
while opcion != "3":
    opcion, monto, pago = operaciones[k]
    k += 1
    if opcion == "1":
        cobros += 1
        if pago == "E":
            efectivo += monto
        else:
            tarjeta += monto
        print(f"Cobro #{cobros}: ${monto:,.2f} ({pago})")
    elif opcion == "2":
        print(f"Parcial -> efectivo ${efectivo:,.2f} | tarjeta ${tarjeta:,.2f}")
total = efectivo + tarjeta
print(f"CORTE: {cobros} cobros, total ${total:,.2f}, promedio ${total / cobros:,.2f}")
````

## Ejercicio 7 — Depuración con PDB (Tema 7)
**Enunciado:** encontrar el error en un cálculo de recargo que da un total incorrecto.

**Modelo EPS**
- **Entrada:** `base = 4125.0`, pago con tarjeta.
- **Proceso:** se esperaba `total = base * 1.03 = 4248.75`, pero el código con error suma `3` en lugar de `3 %`.
- **Salida:** total corregido.

**Procedimiento con PDB:** se coloca `breakpoint()` (equivale a `import pdb; pdb.set_trace()`) antes de la línea sospechosa y se usan los comandos:

| Comando | Uso |
|---|---|
| `n` | ejecutar la siguiente línea |
| `p recargo` | imprimir el valor de una variable |
| `l` | ver el código alrededor |
| `c` | continuar la ejecución |

En la sesión, `p recargo` mostró `3` en lugar de `123.75`, lo que reveló que faltaba multiplicar por la base. A continuación se muestra la versión con error y la corregida (el `breakpoint()` se deja comentado para que el notebook corra sin detenerse).

````python
RECARGO_TARJETA = 0.03
base = 4125.0

# Versión con error
recargo = 3                      # <-- BUG: se usó el porcentaje como monto
# breakpoint()                   # aquí: p recargo  ->  3
print("Con error :", base + recargo)

# Versión corregida
recargo = base * RECARGO_TARJETA
print("Corregido :", round(base + recargo, 2))
````

## Ejercicio 8 — Plan de sesiones con `for` (Tema 8)
**Enunciado:** dado un número conocido de sesiones, imprimir en qué semana cae cada una (2 sesiones por semana).

**Modelo EPS**
- **Entrada:** `sesiones` (int).
- **Proceso:** `for numero in range(1, sesiones + 1)`; `semana = (numero + 1) // 2`.
- **Salida:** listado sesión → semana.

**Procedimiento:** se usa `for` porque el número de repeticiones se conoce antes de iniciar el ciclo (a diferencia del `while` del menú).

````python
sesiones = 7
for numero in range(1, sesiones + 1):
    semana = (numero + 1) // 2
    print(f"Sesión {numero} -> semana {semana}")
````

## Ejercicio 9 — Servicio más solicitado con `for` + `if` (Temas 5 y 8)
**Enunciado:** recorrer los cobros del día y determinar qué servicio se pidió más.

**Modelo EPS**
- **Entrada:** códigos de servicio del día (1-4).
- **Proceso:** `for` que recorre los códigos, contadores por servicio y comparación para hallar el máximo.
- **Salida:** nombre del servicio más solicitado y su conteo.

**Procedimiento:** es la misma lógica del corte de caja del prototipo.

````python
servicios_dia = "2132123"     # un dígito por paciente atendido
c1 = c2 = c3 = c4 = 0
for s in servicios_dia:
    if s == "1":
        c1 += 1
    elif s == "2":
        c2 += 1
    elif s == "3":
        c3 += 1
    else:
        c4 += 1

nombre, mayor = "Fisioterapia general", c1
if c2 > mayor:
    nombre, mayor = "Rehabilitación deportiva", c2
if c3 > mayor:
    nombre, mayor = "Terapia de sanación", c3
if c4 > mayor:
    nombre, mayor = "Electroterapia", c4
print(f"Conteos: fisio={c1}, deportiva={c2}, sanación={c3}, electro={c4}")
print(f"Más solicitado: {nombre} ({mayor})")
````
